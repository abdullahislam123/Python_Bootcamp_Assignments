 Example:


Enter a positive number: -3

3. Fibonacci Sequence
 Write a Python program to print the Fibonacci sequence up to a given number of terms.
 Example:
Enter number of terms: 7  
Output: 0 1 1 2 3 5 8


```python
def fibonacci_sequence(n_terms):
    if n_terms <= 0:
        print("Please enter a positive integer.")
    elif n_terms == 1:
        print("Fibonacci sequence up to", n_terms, ":")
        print(0)
    else:
        print("Fibonacci sequence up to", n_terms, ":")
        a, b = 0, 1
        count = 0
        while count < n_terms:
            print(a, end=" ")
            nth = a + b
            a = b
            b = nth
            count += 1
        print()

while True:
    try:
        num_terms = int(input("Enter number of terms: "))
        if num_terms >= 0:
            break
        else:
            print("Please enter a non-negative number.")
    except ValueError:
        print("Invalid input. Please enter an integer.")

fibonacci_sequence(num_terms)
```

5. Even and Odd Counter
 Write a Python program that takes a list of numbers as input from the user and counts how many numbers are even and how many are odd.

```python
def count_even_odd(numbers):
    even_count = 0
    odd_count = 0
    for num in numbers:
        if num % 2 == 0:
            even_count += 1
        else:
            odd_count += 1
    return even_count, odd_count

while True:
    try:
        input_str = input("Enter a list of numbers separated by spaces: ")
        numbers = [int(x) for x in input_str.split()]
        break
    except ValueError:
        print("Invalid input. Please enter a list of integers separated by spaces.")

even_count, odd_count = count_even_odd(numbers)

print(f"Number of even numbers: {even_count}")
print(f"Number of odd numbers: {odd_count}")
```

6. Factorial Finder
 Write a Python program to calculate the factorial of a number using a loop.
 If the number is negative, display an appropriate message.

```python
def factorial(n):
    if n < 0:
        return "Factorial is not defined for negative numbers"
    elif n == 0:
        return 1
    else:
        result = 1
        for i in range(1, n + 1):
            result *= i
        return result

while True:
    try:
        num = int(input("Enter a number to find its factorial: "))
        break
    except ValueError:
        print("Invalid input. Please enter an integer.")

print(f"The factorial of {num} is: {factorial(num)}")
```

7. Find Maximum and Minimum in a List
 Write a Python program that takes a list of integers from the user and prints the maximum and minimum numbers without using the built-in max() or min() functions.

```python
def find_max_min(numbers):
    if not numbers:
        return None, None  # Handle empty list case

    maximum = numbers[0]
    minimum = numbers[0]

    for num in numbers:
        if num > maximum:
            maximum = num
        if num < minimum:
            minimum = num
    return maximum, minimum

while True:
    try:
        input_str = input("Enter a list of integers separated by spaces: ")
        numbers = [int(x) for x in input_str.split()]
        if numbers:
            break
        else:
            print("Please enter at least one integer.")
    except ValueError:
        print("Invalid input. Please enter a list of integers separated by spaces.")

maximum, minimum = find_max_min(numbers)

if maximum is not None and minimum is not None:
    print(f"Maximum number: {maximum}")
    print(f"Minimum number: {minimum}")
```

8. Web Scraping – Extract Titles
 Using the requests and BeautifulSoup libraries, write a Python program to:
Fetch the HTML of a webpage (e.g., https://quotes.toscrape.com)


Extract and print all the quote texts and authors’ names from the page.

```python
import requests
from bs4 import BeautifulSoup

url = "https://quotes.toscrape.com"
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')

quotes = soup.find_all('div', class_='quote')

for quote in quotes:
    text = quote.find('span', class_='text').text
    author = quote.find('small', class_='author').text
    print(f"Quote: {text}")
    print(f"Author: {author}")
    print("-" * 20)
```

9. Web Scraping – Save Data to CSV
 Write a Python script that scrapes article titles and links from a website (e.g., https://news.ycombinator.com) and saves them into a CSV file with columns:
 Title, Link


```python
import requests
from bs4 import BeautifulSoup
import csv

url = "https://news.ycombinator.com"
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')

articles = soup.find_all('tr', class_='athing')

with open('hacker_news.csv', 'w', newline='', encoding='utf-8') as csvfile:
    csv_writer = csv.writer(csvfile)
    csv_writer.writerow(['Title', 'Link'])  # Write header

    for article in articles:
        title_tag = article.find('span', class_='titleline').find('a')
        title = title_tag.text
        link = title_tag['href']
        csv_writer.writerow([title, link])

print("Data scraped and saved to hacker_news.csv")
```
